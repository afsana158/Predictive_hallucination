```python
import numpy as np

# FEVER does not have a real candidate answer,
# so question-answer similarity is not applicable.
train_df.loc[
    train_df["source_dataset"] == "fever",
    "question_answer_similarity"
] = np.nan

validation_df.loc[
    validation_df["source_dataset"] == "fever",
    "question_answer_similarity"
] = np.nan

test_df.loc[
    test_df["source_dataset"] == "fever",
    "question_answer_similarity"
] = np.nan
```

```python
# Context-based similarity is not applicable
# when no actual context is available.

for df in [train_df, validation_df, test_df]:

    no_context = df["has_context"] == 0

    df.loc[
        no_context,
        "question_context_similarity"
    ] = np.nan

    df.loc[
        no_context,
        "answer_context_similarity"
    ] = np.nan
```

```python
halueval_train = pd.read_parquet(
    "../data/processed/halueval_train.parquet"
)
```

```python
print(
    halueval_train[
        ["question", "context"]
    ].head()
)
```

```python
retrieval_corpus = (
    halueval_train[
        ["context"]
    ]
    .dropna()
    .drop_duplicates()
    .rename(
        columns={"context": "document"}
    )
    .reset_index(drop=True)
)
```

```python
print("Corpus size:", len(retrieval_corpus))
print(retrieval_corpus.head())
```

```python
retrieval_corpus = retrieval_corpus[
    retrieval_corpus["document"]
    .fillna("")
    .str.strip()
    .ne("")
].reset_index(drop=True)

print("Final corpus size:", len(retrieval_corpus))
```

```python
import faiss
import numpy as np
from sentence_transformers import SentenceTransformer
```

```python
embedding_model = SentenceTransformer(
    "all-MiniLM-L6-v2"
)
```

```python
corpus_embeddings = embedding_model.encode(
    retrieval_corpus["document"].tolist(),
    batch_size=64,
    show_progress_bar=True,
    convert_to_numpy=True
)
```

```python
faiss.normalize_L2(corpus_embeddings)
```

```python
dimension = corpus_embeddings.shape[1]

index = faiss.IndexFlatIP(dimension)

index.add(corpus_embeddings)
```

```python
print("Documents indexed:", index.ntotal)
```

```python
query = train_df.iloc[0]["question"]

print("Query:")
print(query)
```

```python
query_embedding = embedding_model.encode(
    [query],
    convert_to_numpy=True
)

faiss.normalize_L2(query_embedding)
```

```python
scores, indices = index.search(
    query_embedding,
    3
)
```

```python
for rank, (score, idx) in enumerate(
    zip(scores[0], indices[0]),
    start=1
):
    print(f"\nRank {rank}")
    print("Score:", score)
    print(
        "Document:",
        retrieval_corpus.iloc[idx]["document"]
    )
```
