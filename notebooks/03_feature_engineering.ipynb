```python
import numpy as np

# FEVER does not have a real candidate answer,
# so question-answer similarity is not applicable.
train_df.loc[
    train_df["source_dataset"] == "fever",
    "question_answer_similarity"
] = np.nan

validation_df.loc[
    validation_df["source_dataset"] == "fever",
    "question_answer_similarity"
] = np.nan

test_df.loc[
    test_df["source_dataset"] == "fever",
    "question_answer_similarity"
] = np.nan
```

```python
# Context-based similarity is not applicable
# when no actual context is available.

for df in [train_df, validation_df, test_df]:

    no_context = df["has_context"] == 0

    df.loc[
        no_context,
        "question_context_similarity"
    ] = np.nan

    df.loc[
        no_context,
        "answer_context_similarity"
    ] = np.nan
```
